# Getting Started with cvh-client

This notebook demonstrates how to use `cvh-client` to authenticate with Auth0 and make API requests.

## Setup

Make sure you have the package installed:

```bash
pip install --index-url https://test.pypi.org/simple/ --extra-index-url https://pypi.org/simple/ cvh-client
```

## Configuration

Replace these values with your Auth0 application settings and API base URL.

````python
AUTH0_DOMAIN = "your-tenant.auth0.com"
AUTH0_CLIENT_ID = "your-client-id"
API_AUDIENCE = "https://api.example.com"
API_BASE_URL = "https://api.example.com"
````

## Authenticate and create the client

This opens a browser window for Auth0 login. Follow the instructions printed below the cell to complete authentication. The client will automatically re-authenticate if the token expires.

````python
from cvh_client import CVHClient

client = CVHClient.from_login(
    base_url=API_BASE_URL,
    domain=AUTH0_DOMAIN,
    client_id=AUTH0_CLIENT_ID,
    audience=API_AUDIENCE,
)
````

## List workspaces

````python
workspaces = client.list_workspaces()
for ws in workspaces.items:
    print(f"{ws.name} ({ws.uuid}) — {ws.datasets_count} datasets, {ws.visualizations_count} visualizations")
````

````python
# Use the first workspace for the following examples
workspace_uuid = workspaces.items[0].uuid
workspace_uuid
````

## List datasets in a workspace

````python
datasets = client.list_datasets(workspace_uuid)
for ds in datasets.items:
    print(f"{ds.name} ({ds.file_type}, {ds.assembly}) — {ds.uuid}")
````

## Get a single dataset

````python
dataset = client.get_dataset(datasets.items[0].uuid)
dataset
````

## List visualizations in a workspace

````python
visualizations = client.list_visualizations(workspace_uuid)
for viz in visualizations:
    print(f"{viz.name} ({viz.tool}) — {viz.uuid}")
````

## Get a single visualization

````python
viz = client.get_visualization(visualizations[0].uuid)
viz
````

## Update a visualization

````python
client.update_visualization(viz.uuid, name="Renamed Visualization", description="Updated via cvh-client")
````

## Update a dataset

````python
client.update_dataset(dataset.uuid, name="Renamed Dataset")
````
